```python
import pyvips
import openslide
import pandas as pd
```

```python
data = pd.read_csv('../slide_mpp_info.csv')
data
```

```python
filtered = data[(data['resolution'].isna() & data['level'].isna() & data['mpp_x'].isna() & data['mpp_y'].isna() & data['level_downsamples'].isna())| ((data['resolution'] == 20.0) & (data['mpp_x'] < 0.4))]
filtered
```

```python
all_na = data[(data['resolution'].isna() & data['level'].isna() & data['mpp_x'].isna() & data['mpp_y'].isna() & data['level_downsamples'].isna())| ((data['resolution'] == 20.0) & (data['mpp_x'] < 0.4))]
all_na
```

```python
slide = openslide.OpenSlide(f'../dataset/{filtered.iloc[-2]["slide_name"]}')
slide.properties
```

```python
image = pyvips.Image.new_from_file(f'../dataset/{filtered.iloc[0]["slide_name"]}', access="sequential")

print("=== Metadata disponibili ===")
for key in image.get_fields():
    try:
        print(f"{key}: {image.get(key)}")
    except:
        pass
```

```python
try:
    mpp_x = float(image.get("openslide.mpp-x"))
    mpp_y = float(image.get("openslide.mpp-y"))
    print(f"\nMicrons per pixel (MPP): {mpp_x} x {mpp_y}")
except:
    print("\nMicrons per pixel (MPP) non trovati.")
```

```python
try:
    magnification = image.get("aperio.AppMag")
    print(f"Magnification: {magnification}x")
except:
    print("Magnification non trovata.")
```
